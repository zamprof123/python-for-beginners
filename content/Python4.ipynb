## What is pandas?

pandas is a Python library for working with tables of data.

A table in pandas is called a **DataFrame**:

 - columns have names

 - each column has a type (numbers, text, True/False)

 - each row is one observation

# Comparing groups

Let's try comparing groups. First, a **box plot**

```python
df.boxplot(column="score", by="sex")
plt.title("Scores by sex")
plt.suptitle("")
plt.ylabel("Score")
plt.show()
```

Do you think there is a significant difference?

The box shows where the middle 50% of the data lies. If you ordered all the scores (for one or the other, male or female), then mark the points where you are 25% along, 50% and 75%, these are the quartiles (the 50% quartile is also called the median). The 25% and 75% quartiles are the bottom and top of the box, and the median is the green line. 

this gives you a visual way to look at how the scores are distributed.

# Let's import a file that contains 13 different datasets!

Each dataset consists of a lot of (x,y) coordinates.

The file format is going to be this. On each line there are columns. 
The first column tells you the dataset name; 
the second and third columns tell you the x and y coordinates
of a datapoint in that dataset.

```python
data = pd.read_csv('Thirteen.tsv', sep='\t')
data.head()
```

```python
#How many datasets?
len(data['dataset'].unique())
```

# Let's look at the distributions of the data in each of these datasets!
We use the pandas groupby function, to group the data in that dataframe by whatever their dataset name is.
Then let's calculate the number of datapoints, the mean, variance and standard deviation of each

```python
datasets_grouped = data.groupby('dataset')
datasets_grouped.agg(['count', 'mean', 'var', 'std'])
```

ooh ok. 
**All the datasets are almost the same surely - they all have almost the same mean and variance and standard deviation.**

# Let's plot each dataset on a big grid of plots and look at them all!

```python
#i'm going to use a slightly different plotting library called seaborn for this
import seaborn as sns 

grid_scatterplots = sns.FacetGrid(data, col="dataset", hue="dataset", col_wrap=4)
grid_scatterplots.map_dataframe(sns.scatterplot, x="x", y="y")
```

## This a famous and instructive set of datasets!

Looking at summary statistics like "the mean" is a very limited way of looking at the data.
Much better to *really look*, and plot it. They all have the same mean and variance, but yet are very different.
